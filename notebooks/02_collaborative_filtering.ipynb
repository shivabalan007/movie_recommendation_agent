```python
movie_ratings = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\processed\movie_ratings.csv"
)

movie_ratings.head()
```

```python
user_movie_matrix = movie_ratings.pivot_table(
    index="user_id",
    columns="movie_title",
    values="ratings"
) # Makes normal columns into User-Item Matrix
```

```python
movie_similarity_df = pd.DataFrame(
    movie_similarity,
    index=user_movie_matrix.columns,
    columns=user_movie_matrix.columns
) # can see output of movie similarity
```

**Movie_lens_latest**

```python
ratings = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lens_latest\ratings.csv"
)

movies = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lens_latest\movies.csv"
)
```

```python
ratings.shape
movies.shape

print(ratings.columns)
print(movies.columns)
```

```python
movie_ratings = ratings.merge(
    movies,
    on="movieId"
)
```

```python
print(movie_ratings.shape)
print(movie_ratings.head())
```

```python
movie_user_matrix = movie_ratings.pivot_table(
    index="title",
    columns="userId",
    values="rating"
).fillna(0)
```

```python
print(movie_user_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

movie_similarity = cosine_similarity(
    movie_user_matrix
)
```

```python
similarity_df = pd.DataFrame(
    movie_similarity,
    index=movie_user_matrix.index,
    columns=movie_user_matrix.index
)
```

```python
print(similarity_df.shape)
```

```python
print(
    similarity_df.loc[
        "Interstellar (2014)"
    ].sort_values(ascending=False).head(10)
)
```

```python
def recommend_cf_movies(movie_name, top_n=10):
    similar_movies = ( similarity_df[movie_name]
        .sort_values(ascending=False)
        .iloc[1:top_n+1]
    )
    
    return similar_movies.index.tolist()
```

```python
print(
    recommend_cf_movies(
        "Interstellar (2014)"
    )
)
```
