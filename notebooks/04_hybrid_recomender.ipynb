**Movie_lense_latest**

```python
ratings = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lens_latest\ratings.csv"
)

movies = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lens_latest\movies.csv"
)

tags = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lens_latest\tags.csv"
)


print("Ratings shape:", ratings.shape)
print("Movies shape :", movies.shape)
print("Tags shape   :", tags.shape)
```

```python
titles1 = set(
    latest_movies["title"]
    .dropna()
    .astype(str)
    .str.lower()
    .str.strip()
)

titles2 = set(
    movies["title"]
    .dropna()
    .astype(str)
    .str.replace(r"\(\d{4}\)", "", regex=True)  # remove MovieLens year
    .str.lower()
    .str.strip()
)

overlap = titles1 & titles2

print("latest_movies shape:", latest_movies.shape)
print("ratings shape      :", ratings.shape)
print("movies shape       :", movies.shape)
print("tags shape         :", tags.shape)

print("\nLatest titles      :", len(titles1))
print("MovieLens titles   :", len(titles2))
print("Overlap count      :", len(overlap))
```
