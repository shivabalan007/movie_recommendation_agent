```python
import pandas as pd  #Import Lib
import numpy as np
```

```python
ratings = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lense\u.data",
    sep="\t",
    names=["user_id", "movie_id", "ratings", "timestamp"]
)

ratings.head()
```

```python
movies = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lense\u.item",
    sep="|",
    encoding="latin-1",
    header=None
)

movies.head()
```

```python
movies = movies[[0,1]]   #Keeping only required columns

movies.columns = ['movie_id', "movie_title"]

movies.head()
```

```python
users = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\movie_lense\u.user",
    sep="|",
    names=["user_id", "age", "gender", "occupation", "zip_code"]
)

users.head()
```

```python
latest_movies = pd.read_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\raw\latest_movies_2025\Latest 2025 movies Datasets.csv"
)

latest_movies.head()
```

```python
latest_movies = latest_movies[
    [
        "title",
        "overview",
        "release_date",
        "vote_average",
        "popularity"
    ]
]

latest_movies.head()
```

```python
print(ratings.shape)
print(movies.shape)
print(users.shape)
print(latest_movies.shape)
```

```python
ratings.info()
movies.info()
latest_movies.info()
```

```python
latest_movies.isnull().sum()   #checking Missing value 
```

```python
latest_movies = latest_movies.dropna(subset=["overview"])
#removed all NaN values in overview
```

```python
latest_movies.shape
latest_movies.isnull().sum()
```

```python
latest_movies["release_date"] = pd.to_datetime(
    latest_movies["release_date"],
    errors="coerce"  #Give NaT(Not a Time) to invalid dates
)
```

```python
latest_movies["release_year"] = (
    latest_movies["release_date"].dt.year #new feature
)
```

```python
latest_movies.info()
```

```python
latest_movies[latest_movies['release_date'].isnull()]
#Output of missing columns
```

```python
movie_ratings = pd.merge(
    ratings,
    movies,
    on="movie_id" #merge movie based on movie_id
)

movie_ratings.head()
```

```python
movie_ratings = movie_ratings.drop("timestamp", axis=1)
```

```python
movie_ratings.head()
```

```python
#Most rated movies
#------------------
movie_ratings["movie_title"].value_counts().head(10)
```

```python
#Highest avg-rated movies - this is not fair even one movie with one 5 ratings take the first place 
#------------------
movie_ratings.groupby("movie_title")["ratings"]\
.mean()\
.sort_values(ascending=False)\
.head(10)
```

```python
movie_ratings.to_csv(
    r"C:\Dev\projects\movie_recommemdation_agent\data\processed\movie_ratings.csv",
    index=False
)
```
